```python

graph = {
    0: [1, 2],
    1: [3],
    2: [4],
    3: [],
    4: [5],
    5: []
}

visited = set()

def dfs(node):
    if node not in visited:
        print(node, end=" ")
        visited.add(node)
        for neighbor in graph[node]:
            dfs(neighbor)

print("DFS Traversal:")
dfs(0)
```

```python


def dfs_recursive(G, start, visited=None):
    """
    Recursive DFS functio
    G: adjacency list (dictionary)
    start: starting node
    visited: set to track visited nodes
    """
    if visited is None:
        visited = set() 



    visited.add(start)
    print(start, end=" ")


    for neighbor in G[start]:
        if neighbor not in visited:
            dfs_recursive(G, neighbor, visited)



G = {
    0: [1, 2],
    1: [3],
    2: [4],
    3: [],
    4: [5],
    5: []
}

print("DFS Traversal starting from node 0:")
dfs_recursive(G, 0)
```

```python
def dfs_with_path(G, start, visited=None, path=None):
    if visited is None:
        visited = set()
    if path is None:
        path = []

    visited.add(start)
    path.append(start)

    if len(path) > 1:
        print("Path to", start, ":", " → ".join(map(str, path)))

    for neighbor in G[start]:
        if neighbor not in visited:
            dfs_with_path(G, neighbor, visited, path)

    path.pop()


G = {
    0: [1, 2],
    1: [3],
    2: [4],
    3: [],
    4: [5],
    5: []
}

dfs_with_path(G, 0)
```

```python
from queue import Queue

def bfs(G, start):
    visited = set()
    q = Queue()

    visited.add(start)
    q.put(start)

    while not q.empty():
        node = q.get()
        print(node, end=" ")

        for neighbor in G[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                q.put(neighbor)


G = {
    0: [1, 2],
    1: [3],
    2: [4],
    3: [],
    4: [5],
    5: []
}

print("BFS Traversal:")
bfs(G, 0)
```

```python
from queue import Queue


filename = "g.txt"
with open(filename, "r") as f:
    lines = f.readlines()


n, e = map(int, lines[0].split())


adj_matrix = [[0]*n for _ in range(n)]
adj_list = {i: [] for i in range(n)}


for line in lines[1:]:
    u, v = map(int, line.split())

    adj_matrix[u][v] = 1
    adj_matrix[v][u] = 1
    adj_list[u].append(v)
    adj_list[v].append(u)


degree = [sum(adj_matrix[i]) for i in range(n)]


def dfs_recursive(G, start, visited=None):
    if visited is None:
        visited = set()
    visited.add(start)
    print(start, end=" ")
    for neighbor in G[start]:
        if neighbor not in visited:
            dfs_recursive(G, neighbor, visited)

def bfs(G, start):
    visited = set()
    q = Queue()
    visited.add(start)
    q.put(start)

    while not q.empty():
        node = q.get()
        print(node, end=" ")
        for neighbor in G[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                q.put(neighbor)



print("\nAdjacency Matrix:")
for row in adj_matrix:
    print(row)

print("\nAdjacency List:")
for k, v in adj_list.items():
    print(f"{k}: {v}")

print("\nDegree of each node:")
for i in range(n):
    print(f"Node {i}: Degree = {degree[i]}")

print("\nDFS Traversal starting from node 0:")
dfs_recursive(adj_list, 0)

print("\nBFS Traversal starting from node 0:")
bfs(adj_list, 0)

print("\n\nObservation:")
print("- DFS goes deeper first (follows one path down before backtracking)")
print("- BFS finds shortest path in an unweighted graph")
```

```python

def dfs(G, start, visited):
    visited.add(start)
    for neighbor in G[start]:
        if neighbor not in visited:
            dfs(G, neighbor, visited)


def count_connected_components(G):
    visited = set()
    count = 0

    for node in G:
        if node not in visited:
            count += 1
            dfs(G, node, visited)

    return count


G = {
    0: [1, 2],
    1: [0, 3, 4],
    2: [0, 4],
    3: [1, 5],
    4: [2, 1, 5],
    5: [3, 4],
    6: [],       
    7: []         
}

components = count_connected_components(G)
print(f"Number of connected components: {components}")
```
